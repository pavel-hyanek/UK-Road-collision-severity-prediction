```python
non_countable_collision_columns = ['collision_index','lsoa_of_accident_location', 'local_authority_highway_current', 
                                   'local_authority_highway', 'local_authority_ons_district', 'local_authority_district',
                                    'date', 'latitude', 'longitude', 'collision_index', 'collision_ref_no', 
                                    'location_easting_osgr', 'location_northing_osgr', 'time']                            
                                      

for column in collision_df.columns:
    if column not in non_countable_collision_columns:
        print(collision_df[column].value_counts(dropna=False))
```

### Collision missing values and duplicates

```python
collision_df_nan_count = collision_df.isnull().sum()
print(collision_df_nan_count)
```

```python
missing_value_columns = ['location_easting_osgr', 'location_northing_osgr', 'longitude', 'latitude']
missing_entry_rows_candidate = []
missing_entries_in_same_rows = True

for column in missing_value_columns:
    missing_entry_rows = collision_df.index[collision_df[column].isnull()].tolist()
    print(f'Column name: {column}, rows with a missing entry: {missing_entry_rows}')
    if missing_entry_rows_candidate == []:
        missing_entry_rows_candidate = missing_entry_rows
    elif missing_entry_rows_candidate != missing_entry_rows:
        missing_entries_in_same_rows = False

print(f'All missing entries are in the same rows: {missing_entries_in_same_rows}')
```

##### Thus all the missing values have a value of -1 as explained in the data desctription file except for 53 rows with missing location data.

### Check whether all collisions have corresponding vehicles available

```python
collision_indices = set(collision_df['collision_index'])
vehicle_collision_indices = set(vehicle_df['collision_index'])

if collision_indices == vehicle_collision_indices:
    print('All collisions have corresponding vehicle data')
else:
    print(f'{len(collision_indices - vehicle_collision_indices)}/{len(collision_indices)} collisions have no vehicle data')
```

##### 5897/513801 collisions have no vehicle data
